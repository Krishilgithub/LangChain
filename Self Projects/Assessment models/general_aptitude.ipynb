```python
from langchain_huggingface import HuggingFaceEmbeddings, HuggingFaceEndpoint, ChatHuggingFace
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_community.document_loaders import PyPDFLoader
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

hf_api_key = os.getenv("HUGGINGFACEHUB_ACCESS_TOKEN")
```

```python
loader = PyPDFLoader('data/Krishil Agrawal Resume - ML.pdf')

resume = loader.load()
```

```python
parser1 = StrOutputParser()
parser2 = JsonOutputParser()
```

```python
llm = HuggingFaceEndpoint(
  model = "mistralai/Mistral-7B-Instruct-v0.2",
  task="text-generation",
  huggingfacehub_api_token=hf_api_key,
  temperature=0.5
)

model = ChatHuggingFace(
  llm = llm
)
```

```python
prompt1 = "Assume that you are a interviewer and you have to make {count} {difficulty} aptitude mcqs based on the {technical} stack of the resume. Generate mcqs in pure JSON format without including any types of comment, notes. Give me output in pure JSON format"
```

```python
template1 = PromptTemplate(
  template="Give me 10 aptitude question based on the technical stack {technical}",
  input_variables=['technical']
)
```

```python
chain = template1 | model | parser1
```

```python
chain.invoke({'count': 20, 'difficulty': 'medium'})
```
