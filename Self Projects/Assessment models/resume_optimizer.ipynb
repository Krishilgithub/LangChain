```python
from langchain_huggingface import HuggingFaceEmbeddings, HuggingFaceEndpoint, ChatHuggingFace
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser, JsonOutputParser
from langchain_community.document_loaders import PyPDFLoader
from langchain.chains import LLMChain
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

hf_api_key = os.getenv('HUGGINGFACEHUB_ACCESS_TOKEN_BACKUP')
```

```python
llm = HuggingFaceEndpoint(
  model="mistralai/Mixtral-8x7B-Instruct-v0.1",
  task="text-generation",
  huggingfacehub_api_token=hf_api_key,
  temperature=0.7,
  max_new_tokens=1000,
)

model = ChatHuggingFace(
  llm = llm
)
```

```python
parser1 = StrOutputParser()
parser2 = JsonOutputParser()
```

```python
resume_loader1 = PyPDFLoader('./data/Krishil Agrawal Resume - ML.pdf')
```

```python
resume_text = resume_loader1.load()
```

```python
print(resume_text[0].page_content)
```

```python
resume_analysis_template = """
You are an expert career coach analyzing a resume. Given the following resume text:

{resume_text}

Perform the following tasks:
1. **Extract Key Components**: Identify the main sections (e.g., Summary, Education, Experience, Skills) and list key details (e.g., job roles, skills, achievements).
2. **Provide Feedback**: Evaluate the resume's strengths and weaknesses. Consider clarity, relevance, formatting, keyword usage, and impact for a job.
3. **Suggest Improvements**: Recommend specific changes to improve the resume, such as adding quantifiable achievements, optimizing keywords, or restructuring sections.

Format the output as follows:
**Key Components**:
- [Component 1]: [Details]
- [Component 2]: [Details]

**Feedback**:
- Strengths: [List strengths]
- Weaknesses: [List weaknesses]

**Suggestions**:
- [Suggestion 1]
- [Suggestion 2]
"""
```

```python
resume_analysis_prompt = PromptTemplate(
    input_variables=["resume_text"],
    template=resume_analysis_template
)
```

```python
resume_analysis_chain = LLMChain(llm=llm, prompt=resume_analysis_prompt)
```

```python
result = resume_analysis_chain.run(resume_text=resume_text)
```
