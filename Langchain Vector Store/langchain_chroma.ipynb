```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_core.documents import Document
from langchain.vectorstores import Chroma
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import PromptTemplate
```

```python
from langchain.schema import Document

doc1 = Document(
  page_content="Virat Kohli is a cricketer. He is a captain of the Indian cricket team. He is a right-handed batsman and a right-arm off-break bowler.",
  metadata={"source": "doc1"}
)

doc2 = Document(
  page_content="Rohit Sharma is a cricketer. He is a captain of the Indian cricket team. He is a right-handed batsman and a right-arm off-break bowler.",
  metadata={"source": "doc2"}
)

doc3 = Document(
  page_content="Sachin Tendulkar is a cricketer. He is a captain of the Indian cricket team. He is a right-handed batsman and a right-arm off-break bowler.",
  metadata={"source": "doc3"}
)

doc4 = Document(
  page_content="MS Dhoni is a cricketer. He is a captain of the Indian cricket team. He is a right-handed batsman and a right-arm off-break bowler.",
  metadata={"source": "doc4"}
)

doc5 = Document(
  page_content="Yuvraj Singh is a cricketer. He is a captain of the Indian cricket team. He is a right-handed batsman and a right-arm off-break bowler.",
  metadata={"source": "doc5"}
)
```

```python
docs = [doc1, doc2, doc3, doc4, doc5]
```

```python
from langchain_huggingface import HuggingFaceEmbeddings, HuggingFaceEndpoint
from langchain.vectorstores import Chroma
import os
from dotenv import load_dotenv
load_dotenv()

llm = HuggingFaceEndpoint(
  model="meta-llama/Llama-3.1-8B-Instruct",
  task="text-generation",
  huggingfacehub_api_token=os.getenv("HUGGINGFACEHUB_ACCESS_TOKEN")
)

model = HuggingFaceEmbeddings(
  model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
vectorstore = Chroma(
  embedding_function=model,
  persist_directory="chroma_db",
  collection_name="langchain_chroma"
)
```

```python
pip install sentence-transformers
```

```python
vectorstore.add_documents([doc1, doc2, doc3, doc4, doc5])
```

```python
vectorstore.get(include=["metadatas", "documents", "embeddings"])
```

```python
  #* Search Documents

vectorstore.similarity_search(
  query="Who among this are bowlers?",
  k=2     # states the number of documents to return
)
```

```python
#* Search with similarity score

vectorstore.similarity_search_with_score(
  query="Who among this are bowlers?",
  k=2
)

#* It will return the distance between the query and the documents, the lower the distance, the more similar the document is to the query
```

```python

#* Search with score threshold

vectorstore.similarity_search_with_score(
  query="Who among this are bowlers?",
)
```

```python

#* meta data filtering 

vectorstore.similarity_search_with_score(
  query="Who among this are bowlers?",
  filter={"team": "Chennai Super Kings"}
)
```

```python

#* Update Document

update_doc1 = Document(
    page_content="Virat Kohli is a cricketer. He is a captain of the Indian cricket team. He is a  right-handed batsman and a right-arm off-break bowler.",
    metadata={"source": "doc1"}
)

vectorstore.update_documents(
  ids = ['32d4b60c-da8c-4b99-87ae-8278c5a02180'],
  documents = update_doc1
)
```

```python
# 1. Delete the old document by filter (e.g., by metadata "source")
vectorstore.delete(ids=['32d4b60c-da8c-4b99-87ae-8278c5a02180'])
```
